```python
import math
```

```python
#duong chim bay
def haversine(lat1, lon1, lat2, lon2):
  R = 6371.0  # Bán kính Trái Đất (km)
  phi1, phi2 = math.radians(lat1), math.radians(lat2)
  dphi = math.radians(lat2 - lat1)
  dlambda = math.radians(lon2 - lon1)

  a = (
      math.sin(dphi / 2) ** 2
      + math.cos(phi1) * math.cos(phi2) * math.sin(dlambda / 2) ** 2
  )
  return 2 * R * math.asin(math.sqrt(a))

lat1 = 40.719971
lon1 = -74.010040

lat2 = 40.706718
lon2=-74.012268

print(haversine(lat1=lat1, lat2=lat2, lon1=lon1, lon2=lon2))
```

```python
#manhattan
def manhattan_geo_exact(lat1, lon1, lat2, lon2):
  # Đi theo chiều vĩ độ (Bắc - Nam) trước, sau đó đi theo chiều kinh độ (Đông - Tây)
  lat_dist = haversine(lat1, lon1, lat2, lon1)
  lon_dist = haversine(lat2, lon1, lat2, lon2)
  return lat_dist + lon_dist

lat1 = 40.719971
lon1 = -74.010040

lat2 = 40.706718
lon2=-74.012268

print(manhattan_geo_exact(lat1=lat1, lat2=lat2, lon1=lon1, lon2=lon2))
```
